### Missing Data

```python
# Leave anyways
# Filling missing data mean,median,reasonable value
# Remove
```

```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame({'A':[1,2,np.nan,4], 'B':[5,np.nan,np.nan,8], 'C':[10,20,30,40]}) 
```

```python
df
```

```python
df.dropna() #Delete rows with nan
```

```python
df.dropna(axis = 1) #Delete columns with nan
```

```python
df.dropna(axis = 1, thresh = 2) # Como maximo 2 nan usar un porcentaje
```

```python
df
```

```python
df.fillna(value='Fill Value') #value = 0
```

```python
df['A'] = df['A'].fillna(value = 0)
```

```python
df['B'].fillna(value = df['B'].mean())
```

```python
df.fillna(df.mean())
```
